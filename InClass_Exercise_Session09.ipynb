# Time Complexity Analysis

```python
def alg_1(n, depth=0): # O(2^n) - exponential
    #print(f"n={n}@depth={depth}")
    if n < 2:
        return n
    return alg_1(n - 1, depth+1) + alg_1(n - 2, depth+1)

alg_1(1)
```

```python
def alg_2(n): # O(log n) - logarithmic
    def fd(k):
        if k == 0:
            return 0, 1
        a, b = fd(k // 2)
        c = a * (2 * b - a)
        d = a * a + b * b
        return (d, c + d) if k % 2 else (c, d)
    return fd(n)[0]
```

```python
def alg_3(n):  # O(n^2) - quadratic
    result = 0
    for k in range(n + 1):
        a, b = 0, 1
        for _ in range(k):
            a, b = b, a + b
        result = a
    return result
```

```python
def alg_4(n): # O(n) - linear
    a, b = 0, 1
    for _ in range(n):
        a, b = b, a + b
    return a
```

```python
def alg_5(n): # O(n x log n)
    result = 0
    for k in range(n + 1):
        result = alg_2(k) # O(log n) - logarithmic
    return result
```

## Binet's formula

$$F(n) = \operatorname{round}\left(\frac{\varphi^n}{\sqrt{5}}\right), \quad \varphi = \frac{1+\sqrt{5}}{2}$$

```python
import math
import time

def alg_6(n):  # O(1) - constant
    phi = (1 + math.sqrt(5)) / 2
    return round(phi ** n / math.sqrt(5))
```

```python
ALGORITHMS = {
    "O(1)": alg_6,
    "O(log n)": alg_2,
    "O(n)": alg_4,
    "O(n log n)": alg_5,
    "O(n^2)": alg_3,
    "O(2^n)": alg_1,
}

for n in range(31):
    expected = alg_4(n)
    for name, f in ALGORITHMS.items():
        assert f(n) == expected, f"{name} wrong at n={n}"

# Binet's formula only accurate for small n (i.e. up to n=70)
assert all(alg_6(n) == alg_4(n) for n in range(71))

i = 71 # at n=71 off by 1
print(f"Binet's F({i})={alg_6(i)}, while linear F({i})={alg_4(i)}")
```
