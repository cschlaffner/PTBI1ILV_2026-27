# Time Complexity Analysis

```python
def alg_1(n):
    if n < 2:
        return n
    return alg_1(n - 1) + alg_1(n - 2)
```

```python
def alg_2(n):
    def fd(k):
        if k == 0:
            return 0, 1
        a, b = fd(k // 2)
        c = a * (2 * b - a)
        d = a * a + b * b
        return (d, c + d) if k % 2 else (c, d)
    return fd(n)[0]
```

```python
def alg_3(n):
    result = 0
    for k in range(n + 1):
        a, b = 0, 1
        for _ in range(k):
            a, b = b, a + b
        result = a
    return result
```

```python
def alg_4(n):
    a, b = 0, 1
    for _ in range(n):
        a, b = b, a + b
    return a
```

```python
def alg_5(n):
    result = 0
    for k in range(n + 1):
        result = alg_2(k)
    return result
```
