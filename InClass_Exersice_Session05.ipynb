Largest local sum

```python
def lls(nums: list[float]) -> tuple[float, int, int]:
    best_start = best_end = 0
    best_sum = nums[best_start]

    for start in range(0, len(nums)):
        current_sum = nums[start]
        for end in range(start + 1, len(nums)):
            current_sum += nums[end] # current_sum = current_sum + nums[end]
            if current_sum > best_sum:
                best_sum = current_sum
                best_start = start
                best_end = end

    return best_sum, best_start, best_end

print(lls([-2, 1, -3, 4, -1, 2, 1, -5, 4])) # (6, 3, 6 )

def test1():
    # Testcase regular case
    print("Test 1: ", lls([-2, 1, -3, 4, -1, 2, 1, -5, 4]) == (6, 3, 6))

def test2():
    # Test edgecase with one element
    print("Test 2: ", lls([5]) == (5, 0, 0))

test1()
test2()
```

Largest local sum with one loop only

```python
def lls2(nums: list[float]) -> tuple[float, int, int]:
    # check if nums is not empty otherwise throw an error
    if not nums:
        raise ValueError("Input list cannot be empty")

    best_start = best_end = current_start = 0
    best_sum = current_sum = nums[best_start]

    for i in range(1, len(nums)):
        if current_sum < 0:
            current_sum = nums[i]
            current_start = i
        else:
            current_sum += nums[i]

        if current_sum > best_sum:
            best_sum = current_sum
            best_start = current_start
            best_end = i

    return best_sum, best_start, best_end

def test3():
    # Testcase regular case
    print("Test 1: ", lls2([-2, 1, -3, 4, -1, 2, 1, -5, 4]) == (6, 3, 6))

def test4():
    # Test edgecase with one element
    print("Test 2: ", lls2([5]) == (5, 0, 0))

test3()
test4()
```

```python
def gcd(m, n, depth):
    if n > m:
        print("m > n branch at depth", depth)
        return gcd(n, m, depth + 1)
    elif m % n == 0:
        print ("m % n branch at depth", depth)
        return n
    else:
        print("else branch at depth", depth)
        return gcd(n, m % n, depth + 1)

print(gcd(143, 9555, 0))
```
