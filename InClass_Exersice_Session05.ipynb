Largest local sum

```python
def lls(nums: list[float]) -> tuple[float, int, int]:
    best_start = best_end = 0
    best_sum = nums[best_start]

    for start in range(0, len(nums)):
        current_sum = nums[start]
        for end in range(start + 1, len(nums)):
            current_sum += nums[end] # current_sum = current_sum + nums[end]
            if current_sum > best_sum:
                best_sum = current_sum
                best_start = start
                best_end = end

    return best_sum, best_start, best_end

print(lls([-2, 1, -3, 4, -1, 2, 1, -5, 4])) # (6, 3, 6 )

def test1():
    # Testcase regular case
    print("Test 1: ", lls([-2, 1, -3, 4, -1, 2, 1, -5, 4]) == (6, 3, 6))

def test2():
    # Test edgecase with one element
    print("Test 2: ", lls([5]) == (5, 0, 0))

test1()
test2()
```

Largest local sum with one loop only

```python
def lls2(nums: list[float]) -> tuple[float, int, int]:
    # check if nums is not empty otherwise throw an error
    if not nums:
        raise ValueError("Input list cannot be empty")

    best_start = best_end = current_start = 0
    best_sum = current_sum = nums[best_start]

    for i in range(1, len(nums)):
        if current_sum < 0:
            current_sum = nums[i]
            current_start = i
        else:
            current_sum += nums[i]

        if current_sum > best_sum:
            best_sum = current_sum
            best_start = current_start
            best_end = i

    return best_sum, best_start, best_end

def test3():
    # Testcase regular case
    print("Test 1: ", lls2([-2, 1, -3, 4, -1, 2, 1, -5, 4]) == (6, 3, 6))

def test4():
    # Test edgecase with one element
    print("Test 2: ", lls2([5]) == (5, 0, 0))

test3()
test4()
```

```python
def gcd(m, n, depth):
    if n > m:
        print("m > n branch at depth", depth)
        return gcd(n, m, depth + 1)
    elif m % n == 0:
        print ("m % n branch at depth", depth)
        return n
    else:
        print("else branch at depth", depth)
        return gcd(n, m % n, depth + 1)

print(gcd(143, 9555, 0))
```

Longest common substring

```python
def lcs(s1: str, s2: str) -> dict:
    # you need to set and use the variables best_start1 and best_end1 as indices in s1 and best_start2 
    # and best_end2 as indices in s2 for the longest common substring
    best_start1 = None
    best_end1 = None
    best_start2 = None
    best_end2 = None
    
    for start1 in range(0, len(s1)):
        for end1 in range(start1, len(s1)):
            subseq1 = s1[start1:end1+1]
            for start2 in range(0, len(s2)):
                for end2 in range(start2, len(s2)):
                    subseq2 = s2[start2:end2+1]
                    if subseq1 == subseq2:
                        if best_start1 is None or (end1+1 - start1 > best_end1 - best_start1):
                            best_start1 = start1
                            best_end1 = end1+1
                            best_start2 = start2
                            best_end2 = end2+1

    substring = ""
    if best_start1 is not None:
        substring = s1[best_start1:best_end1]

    return {"seq": substring, "starts": (best_start1, best_start2), "ends": (best_end1, best_end2)}

def test4():
    # Testcase regular case
    print("Test 4: ", lcs("he", "h") == {"seq": "h", "starts": (0, 0), "ends": (1, 1)})

def test5():
    # Testcase regular case
    print("Test 5: ", lcs("hello world", "yellow") == {"seq": "ello", "starts": (1, 1), "ends": (5, 5)})

def test6():
    # Testcase regular case
    print("Test 6: ", lcs("abcde", "abfce") == {"seq": "ab", "starts": (0, 0), "ends": (2, 2)})

def test7():
    # Test edgecase with one element
    print("Test 7: ", lcs("a", "a") == {"seq": "a", "starts": (0, 0), "ends": (1, 1)})

def test8():
    # Test edgecase with no common substring
    print(lcs("abc", "def"))
    print("Test 8: ", lcs("abc", "def") == {"seq": "", "starts": (None, None), "ends": (None, None)})

test4()
test5()
test6()
test7()
test8()
```
