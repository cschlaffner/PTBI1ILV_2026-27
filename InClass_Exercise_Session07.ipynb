## Class definition

```python
class Person:
    def __init__(self, fname, lname):
        self.__firstname = fname
        self.__lastname = lname

    def printname(self):
        print(f"First name: {self.firstname}, Last name: {self.lastname}")
```

#### Using the class Person, create an object for John Doe and print the name

```python
x = Person("John", "Doe")
x.printname()
```
