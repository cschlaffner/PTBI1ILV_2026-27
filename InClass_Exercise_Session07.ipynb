## Class definition

```python
class Person:
    def __init__(self, fname, lname):
        self.__firstname = fname
        self.__lastname = lname

    def printname(self):
        print(f"First name: {self.__firstname}, Last name: {self.__lastname}")
```

#### Using the class Person, create an object for John Doe and print the name

```python
x = Person("John", "Doe")
x.printname() # Person.printname(x)

```

## Inheritance: derive student from person

```python
class Student(Person):
    def __init__(self, fname, lname, degree):
        super().__init__(fname, lname)
        self.__degree = degree        

    def printname(self):
        print(f"{self._Person__lastname}, {self._Person__firstname}: {self.__degree}")

y = Student("Mike", "Schmidt", "MSc OMICS Data Science")
y.printname()
```

```python
class ClassBest(Student):
    def __init__(self, fname, lname, degree, grade):
        super().__init__(fname, lname, degree)
        self.__grade = grade

    def printname(self):
        print(f"{self._Person__lastname}, {self._Person__firstname}: {self._Student__degree}, Grade: {self.__grade}")

z = ClassBest("Anna", "Müller", "MSc OMICS Data Science", 1.0)
z.printname()
```
