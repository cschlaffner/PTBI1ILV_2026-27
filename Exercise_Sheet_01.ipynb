### Problem 5
Write code to display only specific items from a list/array that satisfy
the following conditions:
-   The number must be divisible by five.
-   If the number is greater than or equal to 150, then skip it.
-   If the number is greater than 500, then stop the loop.

Example input:
-   Python: $[12, 75, 150, 182, 145, 525, 50]$

````python
x = [12, 75, 150, 182, 145, 525, 50]

print(type(x))

for i in x:
    if i % 5 == 0 and i < 150:
        print(i)
    elif i > 500:
        break
````

### Problem 6
Write code to count the digits of a given number $n$. The code should
print the result to the console. For example, for $n=75869$ the output
should be 5.

````python
def count_digits(n):
    count = 0
    while n > 0:
        n = n // 10
        count = count + 1
    return count


n = 75869
print(count_digits(n))

n_str = str(n)
print(len(n_str))
````

### Problem 7
Write code to print all prime numbers between any given pair of numbers
$n$ and $m$ to the console. Use as example input $n=25$ and $m=50$.

**NOTE:** A prime number is a natural number greater than 1 that
can only be divided by itself and 1 without remainders. Examples are: 2,
3, 5, 7, 11.

````python
start = 0 # inclusive
stop = 10 # exclustive
step = 1

list(range(start, stop, step))

n = 25
m = 5000

for i in range(n, m+1, 1):
    # check if i is prime iterating over each number from i down to 1
    j = i - 1
    while j > 1:
        if i % j ==0:
            break
        j = j - 1

    if j == 1:
        print(i)
````

````python
def check_prime(n, m):
    for i in range(n, m + 1, 1):
 
        #checks if the number is smaller 2 - 1 is not a prime number we can continue to the next
        if i < 2:
            continue
        #checls if it is any even number except 2 - 2 wld be prime, any other number we can skip
        elif  i % 2 == 0 and i != 2:
            continue
        #here we check the rest
        elif is_prime(i) == False:
            continue
        else:
            print(i)
 
#checks if the current number is divisible by 2 or any number up to its square root
# tests whether  n is a multiple of any integer between 2 and  sqrt {n}
# credits to wikipedia primality test
# faster to only check odd numbers here (start with 3, upto sqrt(num), in steps of 2 = only odd numbers)
def is_prime(num):
    for j in range(3, int(num ** (1/2)) + 1, 2):
        if num % j == 0:
            return False
    return True
 
 
n = 25
m = 50
check_prime(n, m)
````
